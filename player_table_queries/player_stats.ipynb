```python
import duckdb
from dotenv import load_dotenv
import os
import matplotlib.pyplot as plt

load_dotenv()
```

```python
DB_PATH = os.getenv("DUCKDB_PATH")
con = duckdb.connect(DB_PATH)
print("Connected")
```

```python
con.execute("""
    CREATE TABLE IF NOT EXISTS skater_stats (
        skater_stat_id  INTEGER PRIMARY KEY,
        player_id       INTEGER REFERENCES players(player_id),
        season_type     VARCHAR DEFAULT 'Career',  -- Career, Playoffs-Career
        GP              INTEGER,
        G               INTEGER,
        A               INTEGER,
        PTS             INTEGER,
        plus_minus      INTEGER,
        EVG             INTEGER,
        PPG             INTEGER,
        SHG             INTEGER,
        GWG             INTEGER,
        EV_A            INTEGER,
        PP_A            INTEGER,
        SH_A            INTEGER,
        SOG             INTEGER,
        SPCT            REAL,
        FO_PCT          REAL
    )
""")
print("skater_stats created.")
```
